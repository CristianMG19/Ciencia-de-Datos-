```python
#Itemroe -- Funcion que permite recorrer la fila una por una 
import pandas as pd

nombre_columnas = ['FECHA', 'PRECIP', 'EVAP', 'TMAX', 'TMIN']

df = pd.read_csv('25001.txt',
               sep='\s+',
               skiprows=19,
               names=nombre_columnas,
               na_values=['Nulo'],
               encoding='latin1'      #Se usa encoding='latin1' para evitar errores con caracteres en español 
              )

print(f"--- Archivo cargado exitosamente. Total de filas: {len(df)} ---")
```

```python
print("\nInformación general de las columnas (Tipos de datos y valores no nulos):")
df.info()
    
print("\nResumen estadístico (Media, Mínimo, Máximo, Cuartiles):")
print(df.describe())
```

```python
print("\n--- Conteo de registros faltantes por fecha ---")
faltantes_por_fecha = {}

for indice, fila in df.iterrows():
    if fila.isnull().any():
        cantidad_nulos = fila.isnull().sum()
        fecha = fila['FECHA']
        faltantes_por_fecha[fecha] = cantidad_nulos

print(f"Se encontraron {len(faltantes_por_fecha)} fechas con datos faltantes.")
print("Ejemplo de las primeras 5 fechas con nulos:")
for fecha in list(faltantes_por_fecha.keys())[:5]:
    print(f"Fecha: {fecha} | Datos faltantes: {faltantes_por_fecha[fecha]}")
```

```python
print("\n--- Limpieza: Rellenando nulos con la media de la columna ---")
    
    # Se crea una copia para no alterar los datos originales directamente
df_limpio = df.copy()
columnas_numericas = ['PRECIP', 'EVAP', 'TMAX', 'TMIN']
    
for columna in columnas_numericas:
    media_columna = df_limpio[columna].mean()
    df_limpio[columna] = df_limpio[columna].fillna(media_columna)
    print(f"Columna '{columna}': nulos reemplazados con la media -> {media_columna:.2f}")
```
